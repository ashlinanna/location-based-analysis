```python
# ==========================================
# LOCATION-BASED RESTAURANT ANALYSIS
# ==========================================

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv("Dataset .csv")

# Display First 5 Rows
print("FIRST 5 ROWS")
print(df.head())

# ==========================================
# DATASET INFORMATION
# ==========================================

print("\nDATASET SHAPE")
print(df.shape)

print("\nCOLUMN NAMES")
print(df.columns)

print("\nDATASET INFO")
print(df.info())

# ==========================================
# CHECK MISSING VALUES
# ==========================================

print("\nMISSING VALUES")
print(df.isnull().sum())

# ==========================================
# REMOVE MISSING VALUES
# ==========================================

df = df.dropna()

print("\nMISSING VALUES AFTER CLEANING")
print(df.isnull().sum())

# ==========================================
# TOP 10 CITIES BY RESTAURANT COUNT
# ==========================================

city_counts = df['City'].value_counts().head(10)

print("\nTOP 10 CITIES")
print(city_counts)

# ==========================================
# VISUALIZATION - CITY DISTRIBUTION
# ==========================================

plt.figure(figsize=(12,6))

city_counts.plot(kind='bar')

plt.title("Top 10 Cities by Restaurant Count")

plt.xlabel("City")

plt.ylabel("Number of Restaurants")

plt.xticks(rotation=45)

plt.show()

# ==========================================
# AVERAGE RATING BY CITY
# ==========================================

avg_rating = df.groupby('City')['Aggregate rating'].mean()

avg_rating = avg_rating.sort_values(ascending=False).head(10)

print("\nTOP CITIES BY AVERAGE RATING")
print(avg_rating)

# ==========================================
# VISUALIZATION - AVERAGE RATINGS
# ==========================================

plt.figure(figsize=(12,6))

avg_rating.plot(kind='bar')

plt.title("Top Cities by Average Restaurant Rating")

plt.xlabel("City")

plt.ylabel("Average Rating")

plt.xticks(rotation=45)

plt.show()

# ==========================================
# PRICE RANGE ANALYSIS
# ==========================================

price_range = df['Price range'].value_counts()

print("\nPRICE RANGE DISTRIBUTION")
print(price_range)

# ==========================================
# VISUALIZATION - PRICE RANGE
# ==========================================

plt.figure(figsize=(8,5))

sns.countplot(x='Price range', data=df)

plt.title("Restaurant Price Range Distribution")

plt.xlabel("Price Range")

plt.ylabel("Count")

plt.show()

# ==========================================
# TOP CUISINES
# ==========================================

top_cuisines = df['Cuisines'].value_counts().head(10)

print("\nTOP 10 CUISINES")
print(top_cuisines)

# ==========================================
# VISUALIZATION - TOP CUISINES
# ==========================================

plt.figure(figsize=(12,6))

top_cuisines.plot(kind='bar')

plt.title("Top 10 Cuisines")

plt.xlabel("Cuisine")

plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()

# ==========================================
# CONCLUSION
# ==========================================

print("\nLocation-Based Restaurant Analysis Completed Successfully!")
```
